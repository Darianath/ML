\[


2. Для трьох векторів
```python

    [ 1,  0,  2],
    [-2,  1,  1]
```

обчисліть значення квадратичної форми для кожного рядка без циклу.

Для другого пункту використайте `np.einsum()`.

````python
A = np.array([
    [5, 1, 0],
    [1, 4, 2],
    [0, 2, 3]
])

x = np.array([2, -1, 3])

q= x @ A @ x

X = np.array([
    [ 2, -1,  3],
    [ 1,  0,  2],
    [-2,  1,  1]
])

q_all=np.einsum("ij, jk, ik->i", X, A, X)

print(f"{q}\n{q_all}")
````

## Завдання 20. Метод найменших квадратів

Для точок
```python

y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])
```

побудуйте пряму
\[
y \approx a + bx.
\]

1. Сформуйте матрицю ознак зі стовпця одиниць і значень `x`.
2. Знайдіть коефіцієнти `a` і `b` методом найменших квадратів.
3. Обчисліть прогнозні значення `y`.
4. Знайдіть суму квадратів залишків.

Використайте `np.linalg.lstsq()`.

````python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])

X=np.column_stack((np.ones_like(x),x))
c, r,rank, s = np.linalg.lstsq(X, y, rcond=None)

a,b= c
y_pred = X @ c
sse=np.sum((y- y_pred)**2)

print(f"{a}\n{b}\n{y_pred}\n{sse}")
````
